###  เหตุผลที่ลาออก

### 2. rto_mandate  ( วันที่ถูกเรียกเขา office ) กับ Attrition

```python
ct = pd.crosstab(
    df['rto_mandate'],
    df['attrition'],
    normalize='index'
) * 100

ax = ct.plot(
    kind='bar',
    stacked=True, # ทำให้ดูซ้อนทับกัน
    color=['blue', 'red'],
    figsize=(8, 5)
) 

# ใส่ตัวเลขบนแท่ง
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.1f%%',
        label_type='center'
    )
    
plt.title('Attrition by RTO Mandate')
plt.xlabel('RTO Mandate')
plt.ylabel('Percentage (%)')
plt.legend(['Stayed', 'Left'])
ax.set_xticklabels(['Not Required', 'Required'], rotation=0)
plt.savefig("../visualizations/eda_plots/05_attrition_by_rto_mandate.png", dpi=100, bbox_inches="tight")
plt.show()

```

**Insight:**   
พนักงานที่ถูกกำหนดให้เข้าออฟฟิศ (RTO Required) มี Attrition สูงกว่ากลุ่มที่ไม่ถูกกำหนด
- Not Required → Attrition ประมาณ 11.7%  
- Required → Attrition ประมาณ 26.9%  

ดังนั้นกลุ่มที่มี RTO mandate มีอัตราการลาออกมากกว่าสองเท่าของกลุ่มที่ไม่ Required

### 3. Burnout Score กับ Attrition

Left = 5.37


ภาพรวมจึงชี้ว่า พนักงานที่ลาออกมีระดับ Burnout สูงกว่าโดยเฉลี่ย

### 5.Commute (การเดินทาง) กับ Attrition

```python
df['commute_bin'] = pd.cut(
    df['commute_minutes'], 
    bins=[0,15,30,45,60,200],
    labels=['0-15','15-30','30-45','45-60','60+']
    )
commute = df.groupby('commute_bin',observed=True)['attrition'].mean()

ax = commute.plot(kind='bar', 
                  color='pink',
                  rot=0)

ax.bar_label(ax.containers[0], padding=3, fmt='%.2f')
plt.ylim(0,commute.max()*1.1)
plt.title("Attrition Rate by Commute Time")
plt.ylabel("Attrition Rate")
plt.xlabel("Commute Minutes")
plt.savefig("../visualizations/eda_plots/08_attrition_by_commute_time.png", dpi=100, bbox_inches="tight")
plt.show()
```

**Insight:** อัตรา Attrition มีแนวโน้มสูงขึ้นเมื่อเวลาเดินทางเพิ่มขึ้น
อัตรา Attrition:  
0–15 นาที = 15.2%  
15–30 นาที = 17.6%  
30–45 นาที = 18.6%  
45–60 นาที = 18.9%  
60+ นาที = 22.1%  

กลุ่มที่เดินทาง 60 นาทีขึ้นไปมี Attrition สูงที่สุด
